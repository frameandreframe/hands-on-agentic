## 概念导读：LLM API 的三要素

调用一个大语言模型，本质就是一次 HTTPS 请求，只需要三样东西：

| 要素 | 是什么 | 在哪配置 |
|---|---|---|
| `base_url` | 服务地址（OpenAI 兼容接口已成为事实标准） | `.env` 的 `ARK_BASE_URL` |
| `api_key` | 你的身份凭证，**等同密码，绝不外泄** | 环境变量或 `.env` |
| `model` | 模型名（本课程默认 `glm-5.3-flash`） | `.env` 的 `ARK_MODEL` |

参考：火山方舟 OpenAI 兼容接口文档；OpenAI Python 库文档（README quickstart）。

## 密钥安全三铁律

1. 密钥只从**环境变量或 `.env`** 读取，绝不写进代码、notebook、文档。
2. 绝不打印密钥明文，只允许 `masked_key()` 显示掩码。
3. `.env` 永远不进 git（`.gitignore` 首行），提交前自查 `git status`。

下面第一个单元格就是自检——注意输出里只有掩码和布尔值，没有明文。

````python
# 自检：密钥是否就位（只显示掩码与布尔值，绝不打印明文）
import os

from llm_utils import ARK_MODEL, masked_key

print("密钥已加载：", bool(os.environ.get("ARK_API_KEY")))
print("密钥掩码：", masked_key())
print("模型：", ARK_MODEL)
````

````python
# 人生第一次调用：`get_llm_response` 接收字符串，返回字符串
from llm_utils import get_llm_response

reply = get_llm_response(
    "用一句话向零基础同学解释：什么是 agent？",
    system="你是课程助教，回答不超过 40 字。",
)
print(reply)
````

````python
# notebook 内的迷你对话（终端完整版见 chat.py：uv run python chat.py）
from llm_utils import chat

history = [
    {"role": "system", "content": "你是课程陪练小机器人，回答简洁，用中文。"},
    {"role": "user", "content": "你好，用一句话自我介绍"},
]
print("机器人>", chat(history))

# 续一轮：把上一轮回答放回历史，模型就「记得」前面说过什么
history.append({"role": "assistant", "content": chat(history)})
history.append({"role": "user", "content": "我上一句问了什么？"})
print("机器人>", chat(history))
````
